topic3 : Object,identity vs equality

"==" checks whether two values are equal.
"is" checks whether two variables refer to the exact same object in memory.

```python
a=[1,2,3]
b=[1,2,3]
print(a==b)
print(a is b)
```

```python
x=100
y=100
print(x==y)
print(x is y)

p=1000
q=1000
print(p==q)
print(p is q)
```
