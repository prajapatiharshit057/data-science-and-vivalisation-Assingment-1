```python
import numpy as np

sales = np.array(
    [[120, 150, 180], [140, 160, 175], [135, 155, 190], [150, 170, 200]]
)
daily_sales = sales.sum(axis=1)

product_sales = sales.sum(axis=0)
best_day = np.argmax(daily_sales) + 1

print("Daily Sales:", daily_sales)
print("Product Sales:", product_sales)
print("Highest Sales Day:", f"Day {best_day}")
```

```python
weights = np.array([48, 62, 75, 81, 54, 69, 90])
above_70 = weights[weights > 70]
cleaned_weights = np.where(weights < 55, 55, weights)

print("Weights > 70:", above_70)
print("Cleaned Weights:", cleaned_weights)
print("Original Weights:", weights)
```

question 24

```python
quantity = np.array([3, 5, 2, 7])
price = np.array([120, 80, 250, 60])
total_cost = quantity * price
overall_bill = total_cost.sum()
pct_contribution = (total_cost / overall_bill) * 100

print("Product Total Costs:", total_cost)
print("Overall Bill:", overall_bill)
print(
    "Percentage Contributions (%):", np.round(pct_contribution, 2)
)
```

question 25

```python
matrix = np.array([[10, 20, 30, 40], [15, 25, 35, 45], [5, 15, 25, 35]])
row_var = matrix.var(axis=1)
row_std = matrix.std(axis=1)

col_var = matrix.var(axis=0)
col_std = matrix.std(axis=0)

print("Row Variances:", row_var)
print("Row Std Devs:", row_std)
print("Column Variances:", col_var)
print("Column Std Devs:", col_std)
```
